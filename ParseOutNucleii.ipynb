# Use this for standard, 128x128x8 for training purposes

```python
def get_window(my_loc):
    xy_window = 256
    z_window = 64
    limits = img[:,0,:,:].shape
    
    center = [np.floor((a.stop+a.start)/2) for a in my_loc]
    center[0] = np.max([z_window/2, center[0]])
    center[0] = np.min([limits[0]-z_window/2, center[0]])

    center[1] = np.max([xy_window/2, center[1]])
    center[1] = np.min([limits[1]-xy_window/2, center[1]])

    center[2] = np.max([xy_window/2, center[2]])
    center[2] = np.min([limits[2]-xy_window/2, center[2]])
    
    z_min = center[0]-z_window/2
    z_max = center[0]+z_window/2
    
    y_min = center[1]-xy_window/2
    y_max = center[1]+xy_window/2
    
    x_min = center[2]-xy_window/2
    x_max = center[2]+xy_window/2
    
    return img[int(z_min):int(z_max):8, 0, int(y_min):int(y_max), int(x_min):int(x_max)]
```
