```python
import napari
import tifffile
import numpy as np
import pandas as pd
```

```python
viewer = napari.Viewer()
```

```python
classified = tifffile.imread('ClassifiedLabels.tif')
raw = tifffile.imread('Raw.tif')
labels = tifffile.imread('Labels.tif')
cdf = pd.read_csv('Statistics.csv')
```

```python
viewer.add_image(raw, blending='additive')
viewer.add_labels(classified)
viewer.add_labels(labels)
```

```python
viewer.layers[-1].properties = cdf[['SizeType', 'Type', 'Nucleus', 'Hetero', 'Euc', 'EucFraction', 'HeteroFraction', 'CX (pix)', 'CY (pix)', 'CZ (pix)', 'CX (unit)', 'CY (unit)',
       'CZ (unit)', 'Xmin (pix)', 'Ymin (pix)', 'Zmin (pix)', 'Xmax (pix)',
       'Ymax (pix)', 'Zmax (pix)', 'VolBounding (pix)', 'RatioVolbox',
       'Vol (unit)', 'Vol (pix)', 'Surf (unit)', 'Surf (pix)',
       'SurfCorr (pix)', 'Comp (pix)', 'Spher (pix)', 'CompCorr (pix)',
       'SpherCorr (pix)', 'Comp (unit)', 'Spher (unit)', 'CompDiscrete',
       'Ell_MajRad', 'Ell_Elon', 'Ell_Flatness']].to_dict()
```
